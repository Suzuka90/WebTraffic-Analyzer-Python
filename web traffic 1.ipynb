```python
import pandas as pd

# DATI REALISTICI sito web (copia!)
traffico = {
    'pagina': ['Home', 'Shop', 'Blog', 'Contatti', 'Home', 'Shop', 'Blog', 'Prodotto1', 'Prodotto2', 'Home'],
    'clicks': [250, 180, 120, 80, 300, 220, 140, 90, 110, 280],
    'visite': [500, 400, 250, 150, 600, 450, 300, 200, 220, 550],
    'ora': [14, 15, 16, 17, 20, 21, 19, 18, 20, 22]
}

df = pd.DataFrame(traffico)
df.to_csv('web_traffic.csv', index=False)
print("✅ Dataset creato!")
print(df.head())

import matplotlib.pyplot as plt

# GRAFICO 1: Top pagine clicks
top_pagine = df.groupby('pagina')['clicks'].sum().sort_values(ascending=False)
top_pagine.plot(kind='bar', color='skyblue', figsize=(8,5))
plt.title('🏆 TOP Pagine per Click')
plt.ylabel('Click Totali')
plt.xticks(rotation=45)
plt.savefig('top_clicks.png')
plt.show()

# GRAFICO 2: Traffico per ora
traffico_ora = df.groupby('ora')['visite'].sum()
traffico_ora.plot(kind='line', marker='o', color='green', figsize=(8,5))
plt.title('📈 Picchi Traffico Orari')
plt.xlabel('Ora Giorno')
plt.ylabel('Visite')
plt.grid(True)
plt.savefig('traffico_ora.png')
plt.show()

# STATISTICHE
print("📊 RISULTATI:")
print("Pagina TOP:", top_pagine.index[0], top_pagine.iloc[0], "click")
print("Ora picco:", traffico_ora.idxmax(), traffico_ora.max(), "visite")

# Salva risultati
risultati = pd.DataFrame({
    'Pagina': top_pagine.index,
    'Click_Totali': top_pagine.values
})
risultati.to_csv('risultati_traffico.csv', index=False)
print("📁 File pronti: web_traffic.csv + risultati_traffico.csv + 2 PNG")
```
